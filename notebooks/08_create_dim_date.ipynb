# Create Dimension Date Table


# Step 1 — Import libraries

```python
import pandas as pd
```

# Step 2 — Create the date range

```python
dates = pd.date_range(start="2025-01-01", end="2025-12-31", freq="D")
df_date = pd.DataFrame({"date": dates})
```

# Step 3 — Add date attributes

```python
df_date["year"] = df_date["date"].dt.year
df_date["month"] = df_date["date"].dt.month
df_date["day"] = df_date["date"].dt.day

df_date["day_of_week"] = df_date["date"].dt.dayofweek + 1 # +1 because number starts at zero
df_date["day_name"] = df_date["date"].dt.day_name()

df_date["week"] = df_date["date"].dt.isocalendar().week
df_date["quarter"] = df_date["date"].dt.quarter

df_date["is_weekend"] = df_date["day_of_week"].isin([5, 6])
```

# Step 4 — Optional features

```python
df_date["month_name"] = df_date["date"].dt.month_name()
```

# Step 5 — Create a surrogate key

```python
df_date["date_id"] = df_date["date"].dt.strftime("%Y%m%d").astype(int)
```

# Step 6 — Reorder columns

```python
df_date = df_date[
    [
        "date_id",
        "date",
        "year",
        "quarter",
        "month",
        "month_name",
        "week",
        "day",
        "day_of_week",
        "day_name",
        "is_weekend"
    ]
]
```
